Ejercicio 6

```python
palabra = (input('• Introduce una PALABRA . . . ').strip().upper())
vocales = {"A","E","I","O","U","Á","É","Í","Ó","Ú","À","È","Ì","Ò","Ù"}
resultado = 0


for lector in palabra:
    if lector in vocales:
        resultado += 1

cantidad = "VOCAL" if resultado == 1 else "VOCALES"

print("PALABRA:", palabra)
print("TIENE:", resultado, cantidad)
```

Ejercicio 8

```python
palabra = input("Introduce una palabra: ")

palindromo = True
izquierda = 0
derecha = len(palabra) -1


while izquierda < derecha:
    if palabra[izquierda] != palabra[derecha]:
        palindromo = False
        break
    izquierda += 1
    derecha -= 1

if palindromo == False:
    resultado = "No es palindromo"
elif palindromo == True:
    resultado = "Si es palindromo"

print(f"{resultado}")
```
