Ejercicio 3

```python
colors = ["\033[92m", "\033[91m", "\033[97m"]
# size = int(input("Ingresa un numero impar"))
size = 5

for green in range(size):
    x = ""
    for red in range(size):
        if green == red: #El Verde se antepone al Rojo
            x += colors[0] + "*"
        elif green + red == size - 1: #Logica para el "*" rojo
            x += colors[1] + "*"
        else:
            x += colors[2] + " "
    print(x)
```

Ejercicio 4

```python
colors = ["\033[91m", "\033[94m"]
size = 5

esp = " "
esp2 = "   "

for pyr in range(size):
    rombo = ""
    if pyr == 0:
        espacios = " " * (size - 2)
        rombo = colors[0] + "*" * (2 * 0 + 1)
        print(espacios + rombo)
    elif pyr == 1:
        espacios = " " * (size - 3)
        rombo = colors[0] + "*" + esp + colors[1] + "*"
        print(espacios + rombo)
    elif pyr == 2:
        espacios = " " * (size - 4)
        rombo = colors[0] + "*" + esp2 + colors[1] + "*"
        print(espacios + rombo)
    elif pyr == 3:
        espacios = " " * (size - 3)
        rombo = colors[0] + "*" + esp + colors[1] + "*"
        print(espacios + rombo)
    elif pyr == 4:
        espacios = " " * (size - 2)
        rombo = colors[0] + "*" * (2 * 0 + 1)
        print(espacios + rombo)
```

```python
colors = ["\033[92m", "\033[91m", "\033[94m"]
# size = int(input("Ingresa un numero impar"))
size = 5
top = colors[2] + "*" * 7
bot = colors[2] + "*" * 7

print(top)
for green in range(size):
    x = ""
    left = colors[2] + "*"
    right = colors[2] + "*"
    for red in range(size):
        if green == red: #Codigo reciclado xD
            x += colors[1] + "*"
        elif green + red == size - 1:
            x += colors[1] + "*"
        else:
            x += colors[2] + " "
    print(left + x + right)
print(bot)
```

```python
colors = ["\033[91m", "\033[93m", "\033[92m", "\033[96m", "\033[95m"]
altura = 5
top = "      *****"

print(colors[4] + top)
for i in range(altura):
    arcoiris= "***" + " "
    estrellas = " " * (altura - i) + arcoiris
    estrellas2 = " " * (i * altura - i * 3) + arcoiris
    print(colors[i] + estrellas + estrellas2)
```
