```python
productos = [("manzana", 3), ("banana", 2), ("pera", 5)]

nombres_filtrados = [nombre for nombre, precio in productos if precio > 2]

print(f"Mi Lista: {productos}")
print(f"Producto No Mileurista: {nombres_filtrados}")
```
