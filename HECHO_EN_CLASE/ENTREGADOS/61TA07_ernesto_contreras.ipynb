Ejercicio 7

```python
palabra = "PALABRA"
diccionario = {}

for i in range(len(palabra)):
    letra = palabra[i]
    if diccionario.get(letra):
        diccionario[letra] += 1
    else:
        diccionario[letra] = 1

diccionario
```
