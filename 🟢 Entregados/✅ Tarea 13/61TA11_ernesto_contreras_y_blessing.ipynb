Ejercicio 1

```python
lista = [10, 25, 84, 73, 91, 168, 22, 252, 300, 336, 77, 8]

multiplos_encontrados = list(filter(lambda numero: numero % 84 == 0, lista))

if len(multiplos_encontrados) == 4:
    print("El hechizo se ha activado!")
else:
    print("El hechizo ha fallado!")

print(multiplos_encontrados)
```

Ejercicio 2

```python
frutos = [3, 8, 12, 4, 11, 10, 6, 9]

par = list(filter(lambda fruto: fruto % 2 == 0, frutos))
menor = list(filter(lambda fruto: fruto <= 10, par))
duplicar = list(map(lambda fruto: fruto * 2, menor))

print(sum(duplicar))
```
