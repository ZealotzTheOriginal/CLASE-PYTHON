Ejercicio 5

```python
def contar_letras(palabra):
    contador = 0
    for letra in palabra:
        contador += 1
    return contador

contar_letras("HOLA")
```

Ejercicio 10

```python
def calculadora(a,b,operacion):
    if operacion == "suma":
        return (a + b)
    elif operacion == "resta":
        return (a - b)
    elif operacion == "multiplicacion":
        return (a * b)
    elif operacion == "division":
        return (a / b)
    else:
        return "Operación no válida"

calculadora(55,2,"division")
```
