```python
### Ejercicio 1 ###

user_number = (input('• Introduce un NUMERO . . . ').strip())

number = int(user_number)
if number > 10:
    result = "• NÚMERO GRANDE"
else:
    result = "• NÚMERO MENOR DE 10"

print(f"{result}")
```

```python
### Ejercicio 2 ###

positive_negative = (input('• Introduce un NUMERO . . . ').strip())

try:
    number = int(positive_negative)

    if number > 0:
        result = "• POSITIVO"
    else:
        result = "• NEGATIVO"

except ValueError:
    result = ("• Entrada vacía ⚠️ Debes introducir un número." if not positive_negative else
              "• Letras detectadas ⚠️ Debes introducir un número, no letras." if positive_negative.isalpha() else
              "• Entrada inválida ⚠️ Por favor, introduce un número válido.")

print(f"{result}")
```

```python
### Ejercicio 3 ###

user_notes = (input('• Introduce el NUMERO de tus NOTAS . . . ').strip())

try:
    number = int(user_notes)

    if number < 5:
        result = "• SUSPENSO"
    elif number <= 6:
        result = "• APROBADO"
    elif number <= 8:
        result = "• NOTABLE"
    elif number <= 10:
        result = "• EXCELENTE"
    else:
        result = "• NOTA INVÁLIDA"

except ValueError:
    result = ("• Entrada vacía ⚠️ Debes introducir el número de tus notas." if not user_notes else
              "• Letras detectadas ⚠️ Debes introducir tus notas en números." if user_notes.isalpha() else
              "• Entrada inválida ⚠️ Por favor, introduce un número válido.")

print(f"{result}")
```

```python
### Ejercicio 4 ###

user_number = (input('• Introduce un NUMERO . . . ').strip())

try:
    number = int(user_number)

    if number > 0:
        if number % 2 == 0:
            result = f"• EL NÚMERO {number} ES PAR."
        else:
            result = f"• EL NÚMERO {number} ES IMPAR."
    else:
        result = "• Entrada inválida ⚠️ Debes introducir un número POSITIVO."

except ValueError:
    result = ("• Entrada vacía ⚠️ Debes introducir un número." if not user_number else
              "• Letras detectadas ⚠️ Debes introducir un número, no letras." if user_number.isalpha() else
              "• Entrada inválida ⚠️ Por favor, introduce un número válido.")

print(f"{result}")
```

```python
### Ejercicio 5 ###

user_age = (input('• Introduce una EDAD . . . ').strip())

try:
    number = int(user_age)

    if number < 12:
        result = "• NIÑO"
    elif number <= 17:
        result = "• ADOLESCENTE"
    elif number <= 64:
        result = "• ADULTO"
    elif number >= 65:
        result = "• JUBILADO"

except ValueError:
    result = ("• Entrada vacía ⚠️ Debes introducir un número." if not user_age else
              "• Letras detectadas ⚠️ Debes introducir un número, no letras." if user_age.isalpha() else
              "• Entrada inválida ⚠️ Por favor, introduce un número válido.")

print(f"{result}")
```

```python
### Ejercicio 6 ###

user_number = (input('• Introduce un NUMERO . . . ').strip())

try:
    number = int(user_number)

    if number < 100:
        result = "• NUMERO MENOR DE 100 Y 200"
    elif number <= 200:
        if number % 5 == 0:
            result = f"• EL NÚMERO {number} ES DIVISIBLE POR 5."
        else:
            result = f"• EL NÚMERO {number} NO ES DIVISIBLE POR 5."
    else:
        result = "• Entrada inválida ⚠️ Debes introducir un número entre 100 y 200."

except ValueError:
    result = ("• Entrada vacía ⚠️ Debes introducir un número." if not user_number else
              "• Letras detectadas ⚠️ Debes introducir un número, no letras." if user_number.isalpha() else
              "• Entrada inválida ⚠️ Por favor, introduce un número válido.")

print(f"{result}")
```

```python
### Ejercicio 8 ###

is_adult = (input('• Introduce una EDAD . . . ').strip())

try:
    number = int(is_adult)
    result = "• PUEDE CONDUCIR" if number >= 18 else "• NO PUEDE CONDUCIR"

except ValueError:
    result = ("• Entrada vacía ⚠️ Debes introducir alguna EDAD en años." if not is_adult else
              "• Letras detectadas ⚠️ Debes introducir una EDAD en números, no letras." if is_adult.isalpha() else
              "• Entrada inválida ⚠️ Por favor, introduce una EDAD válida.")

print(f"{result}")
```

```python
### Ejercicio 9 ###

week_day = (input('• Introduce un DÍA DE LA SEMANA . . . ').strip().upper())
work_days = {"LUNES", "MARTES", "MIÉRCOLES", "MIERCOLES"}
weekend = {"SÁBADO","SABADO", "DOMINGO"}

if week_day in work_days:
    result = "• DÍA LABORABLE"
elif week_day in weekend:
    result = "• FIN DE SEMANA"
else:
    result = "• DÍA NO VALIDO"

print(f"{result}")
```

```python
### Ejercicio 10 ###

num_1 = (input('• Introduce un PRIMER NUMERO . . . ').strip())
num_2 = (input('• Introduce un SEGUNDO NUMERO . . . ').strip())

try:
    n_1 = int(num_1)
    n_2 = int(num_2)

    if n_1 > 0 and n_2 > 0:
        result = f"{n_1 if n_1 > n_2 else n_2}"
    else:
        result = "• Entrada inválida ⚠️ introduce números positivos."

except ValueError:
    result = ("• Entrada vacía ⚠️ Debes introducir números." if not num_1 and not num_2 else
              "• Letras detectadas ⚠️ Debes introducir números, no letras." if num_1.isalpha() and num_2.isalpha() else
              "• Entrada inválida ⚠️ Por favor, introduce números válidos.")

print(f"{result}")
```
