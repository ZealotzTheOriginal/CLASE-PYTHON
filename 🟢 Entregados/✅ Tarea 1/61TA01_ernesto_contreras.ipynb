```python
# Antes de nada quiero mencionar que estaba experimentando y por consecuencia cree codigo extra fuera de las instrucciones.
#
##
### Variable Name (Nombre de Usuario) ###

name = input('• Introduce tu NOMBRE Y APELLIDO . . . ').upper() #Valor Inicial de "Name"

#Filtro de letras
if name.isalpha():
    name = name

#Filtro de digitos
elif name.isdigit():
    name = "DIGITO DETECTADO              ⚠️ TU NOMBRE NO PUEDE ESTAR ESCRITO CON NUMEROS"

#Filtro de Ø
elif name == "":
    name = "N/A                           ⚠️ INTENTA INGRESAR TU NOMBRE Y APELLIDO"

#Filtro de letras y digitos
elif name.isalnum() and not name.isdigit() and not name.isalpha():
    name = "DIGITO Y LETRA DETECTADO      ⚠️ NO PUEDES TENER UN NOMBRE CON NUMEROS Y LETRAS"



#
##
### Variable City (Ciudad del Usuario)

city = input('• Introduce tu CIUDAD ACTUAL . . . ').upper().strip() #Valor Inicial de "City"

#Filtro de letras
if city.isalpha():
    city = city

#Filtro de digitos
elif city.isdigit():
    city = "DIGITO DETECTADO              ⚠️ TU CIUDAD NO PUEDE ESTAR ESCRITA CON NUMEROS"

#Filtro de Ø
elif city == "":
    city = "N/A                           ⚠️ INTENTA INGRESAR EL NOMBRE DE TU CIUDAD ACTUAL"

#Filtro de letras y digitos
elif city.isalnum() and not city.isdigit() and not city.isalpha():
    city = "DIGITO Y LETRA DETECTADO      ⚠️ TU CIUDAD NO PUEDE TENER UN NOMBRE CON NUMEROS Y LETRAS"



#
##
### Variable Pet (Mascota del Usuario + Filtro Booleano)

pet = input('• Tienes una MASCOTA? SI / NO . . .').upper().strip() #Valor Inicial de "Pet"
has_pet = None

#Filtro de letras
if pet.isalpha():
    #Filtro de Respuestas afirmativas
    if pet == "SI" or pet == "SÍ" or pet == "S": 
        has_pet = True

    #Filtro de Respuestas negativas
    elif pet == "NO" or pet == "N" or pet == "O":
        has_pet = False

#Filtro de digitos
elif pet.isdigit():
    pet = "DIGITO DETECTADO              ⚠️ INTENTA USAR 'SI / NO' COMO RESPUESTA"

#Filtro de letras y digitos
elif pet.isalnum() and not pet.isdigit() and not pet.isalpha():
    pet = "DIGITO Y LETRA DETECTADO      ⚠️ INTENTA USAR 'SI / NO' COMO RESPUESTA"

#Filtro de Ø
elif pet == "":
    pet = "N/A                           ⚠️ INTENTA INGRESAR UNA RESPUESTA COMO SI / NO"

if has_pet == False:
    pet = "NO TIENE"

elif has_pet == True:
            
    pet = input('• Introduce el NOMBRE DE TU MASCOTA . . .').upper().strip() #Valor Secundario de "Pet"

    #Filtro de digitos
    if pet.isdigit():
        pet = "DIGITO DETECTADO              ⚠️ TU MASCOTA NO PUEDE TENER UN NOMBRE CON NUMEROS"

    #Filtro de letras y digitos
    elif pet.isalnum() and not pet.isdigit() and not pet.isalpha():
        pet = "DIGITO Y LETRA DETECTADO      ⚠️ TU MASCOTA NO PUEDE TENER UN NOMBRE CON NUMEROS Y LETRAS"

    #Filtro de Ø
    elif pet == "":
        pet = "N/A                           ⚠️ INTENTA INGRESAR EL NOMBRE DE TU MASCOTA"



#
##
### Variable Color (Color Favorito del Usuario)

color = input('• Introduce tu COLOR FAVORITO . . . ').upper().strip() #Valor Inicial de "Color"

#Filtro de letras
if color.isalpha():
    color = color

#Filtro de digitos
elif color.isdigit():
    color = "DIGITO DETECTADO              ⚠️ LOS COLORES SOLO PUEDEN SER ESCRITOS CON LETRAS"

#Filtro de Ø
elif color == "":
    color = "N/A                           ⚠️ INTENTA INGRESAR EL NOMBRE DE UN COLOR"

#Filtro de letras y digitos
elif color.isalnum() and not color.isdigit() and not color.isalpha():
    color = "DIGITO Y LETRA DETECTADO      ⚠️ TU COLOR NO PUEDE SER ESCRITO CON NUMEROS Y LETRAS"



#
##
### Variable Ages (Años del Usuario, Categoria de Edad) ###

user_age = (input('• Introduce tu AÑO DE NACIMIENTO . . . ').upper().strip()) #Valor Inicial de "Ages"

age_category = user_age
ages = user_age

# Filtro de digitos + calculo de categorias
if user_age.isdigit():
        user_age = int(user_age) #Conversion a enteros
        ages = 2025 - user_age

        # Semi-filtro para una edad de un año superior a 2025
        if ages < 0:
            ages = "VIAJERO DEL TIEMPO"
            name = "MARTY MCFLY"
            age_category = "JOVEN"
            city = "HILL VALLEY"
            pet = "EINSTEIN"
            color = "ROJO"

        # Semi-filtro para una edad inferior o igual a 18 años
        elif ages <= 18:
            age_category = 'JOVEN'

        # Semi-filtro para una edad inferior o igual a 40 años
        elif ages <= 40:
            age_category = 'ADULTO'

        # Semi-filtro para una edad inferior o igual a 41 años
        elif ages <= 122:
            age_category = 'MAYOR'

        # Semi-filtro para una edad demasiado alta
        elif    ages >= 123:
            ages = "POSIBLE MENTIROSO             ⚠️ TU EDAD NO PUEDE SUPERAR 122 AÑOS"
            name = "MISTER MENTIROSO              ⚠️ PENITENCIA POR NO PENSAR BIEN"
            age_category = "DINOSAURIO                    ⚠️ FELICIDADES VIVISTE MAS QUE LA PERSONA MAS LONGEVA"

#Filtro de letras
elif user_age.isalpha():
    ages = "FECHA NO DETECTADA            ⚠️ TU AÑO DE NACIMIENTO NO PUEDE SER ESCRITO CON LETRAS"
    age_category = "EDAD NO DETECTADA             ⚠️ TU EDAD NO PUEDE SER ESCRITA CON LETRAS"

#Filtro de letras y digitos
elif user_age.isalnum() and not user_age.isdigit() and not user_age.isalpha():
    ages = "DIGITO Y LETRA DETECTADO      ⚠️ TU EDAD NO PUEDE ESTAR ESCRITA EN NUMEROS Y LETRAS"
    age_category = "DIGITO Y LETRA DETECTADO      ⚠️ TU EDAD NO PUEDE ESTAR ESCRITA EN NUMEROS Y LETRAS"

#Filtro de Ø
elif user_age == "":
    ages = "N/A                           ⚠️ INTENTA INGRESAR UN AÑO DE NACIMIENTO"
    age_category = "N/A                           ⚠️ INTENTA INGRESAR UN AÑO DE NACIMIENTO"
    


#
##
### Informacion para el usuario
advice = "⚠️"
advice_trim = "▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀"

advice_input = (advice in name or advice in city or advice in pet or advice in color or advice in age_category)
if advice_input == True:
    advice
    advice = "Errores Cometidos:"
    advice_trim = advice_trim
if advice_input == False:
    advice = " "
    advice_trim = " "



#
##
### Ficha de Datos

print(f"           Fichero"
      f"\n▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀{advice_trim}\n"
      f"         Creado por\n"
      f"   🦆 Дмитрий & Ernest 😎                                   {advice}\n"
      f"\n"
      f"        NOMBRE • {name}\n"
      f"          AÑOS • {ages}\n"
      f"        CIUDAD • {city}\n"
      f"       MASCOTA • {pet}\n"
      f"         COLOR • {color}\n"
      f"          EDAD • {age_category}\n"
      f"\n"
      f"\n▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀▀{advice_trim}\n")


```
