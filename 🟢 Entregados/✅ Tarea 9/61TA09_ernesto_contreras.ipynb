```python
contactos = {"NOMBRE": ["EMAIL","TELEFONO"],
             "ANA": ["ANA@EJEMPLO.COM", "111-222-333"],
             "CARLOS": ["CARLOS@EJEMPLO.COM", "444-555-666"]}

salir = False
print(f"1. Añadir Contacto\n"
"2. Buscar Contacto\n"
"3. Modificar Contacto\n"
"4. Borrar Contacto\n"
"5. Lista de Contactos\n"
 "6. Salir\n")

while not salir:
    agenda = input('• Bienvenido a la Agenda "Introduce un numero para continuar" . . . ')

    if agenda == "1":
        # AÑADIR
        nombre_añadir = input('• Introduce el NOMBRE del contacto para añadir. . .').upper()
        if nombre_añadir in contactos:
            print(f"{nombre_añadir}' ya existe! Usa 3 para modificarlo.")
            input("Presiona ENTER para continuar...")
        else:
            email_añadir = input(f'• Introduce el email de {nombre_añadir}. . .').upper()
            telefono_añadir = input(f'• Introduce el teléfono de {nombre_añadir}. . .').upper()
            contactos[nombre_añadir] = [email_añadir, telefono_añadir]
            
            print(f"¡Contacto '{nombre_añadir}' añadido con éxito!")
            input("Presiona Enter para continuar...")

    elif agenda == "2":
        # LISTAR
        print("Lista de Contactos")
        if not contactos:
            print("¡La agenda está vacía! Añade algunos contactos primero.")
        else:
            for i, nombre_contacto in enumerate(contactos):
                datos_contacto = contactos[nombre_contacto]
                email = datos_contacto[0]
                telefono = datos_contacto[1]
                print(f"{i+1}. Nombre: {nombre_contacto}, Email: {email}, Teléfono: {telefono}")
        input("Presiona Enter para continuar...")

    elif agenda == "3":
        # BUSCAR
        if not contactos:
            print("¡La agenda está vacía! No hay contactos para buscar.\n")
        else:
            termino_busqueda = input('• Introduce el NOMBRE, EMAIL o TELEFONO a buscar. . .').upper()
            encontrado = False 
            for nombre_contacto, datos_contacto in contactos.items():
                email = datos_contacto[0]
                telefono = datos_contacto[1]
                if (termino_busqueda == nombre_contacto or 
                    termino_busqueda == email or 
                    termino_busqueda == telefono):
                    print(f"Nombre: {nombre_contacto}")
                    print(f"Email: {email}")
                    print(f"Teléfono: {telefono}\n")
                    encontrado = True
            if not encontrado:
                print(f"\n¡No se encontró ningún contacto con '{termino_busqueda}' en la agenda!\n")
        input("Presiona Enter para continuar...")
    
    elif agenda == "4":
        # MODIFICAR
            if not contactos:
                print("¡La agenda está vacía! No hay contactos para modificar.")
            else:
                termino_busqueda = input('• Introduce el NOMBRE, EMAIL o TELEFONO del contacto a MODIFICAR. . .').upper()
                
                contacto_encontrado_key = None
                datos_contacto_originales = None

                for nombre_contacto, datos_contacto in contactos.items():
                    email = datos_contacto[0]
                    telefono = datos_contacto[1]

                    if (termino_busqueda == nombre_contacto or
                        termino_busqueda == email or
                        termino_busqueda == telefono):
                        contacto_encontrado_key = nombre_contacto
                        datos_contacto_originales = datos_contacto[:] # Hacemos una copia de los datos
                        break

                if contacto_encontrado_key:
                    print(f"\nContacto encontrado: {contacto_encontrado_key}")
                    print(f"  Email actual: {datos_contacto_originales[0]}")
                    print(f"  Teléfono actual: {datos_contacto_originales[1]}\n")

                    while True:
                        opcion_modificar = input("• Que quieres modificar: 1. Nombre / 2. Email / 3. Telefono / 4. Salir ")

                        if opcion_modificar == "1":
                            nuevo_nombre = input("• Introduce el NUEVO NOMBRE: ").upper()
                            if nuevo_nombre in contactos and nuevo_nombre != contacto_encontrado_key:
                                print("¡Error! Ya existe un contacto con ese nombre. Intenta con otro.\n")
                            else:
                                # Creamos una nueva entrada con el nuevo nombre y los datos existentes
                                contactos[nuevo_nombre] = datos_contacto_originales
                                # Eliminamos la entrada antigua
                                del contactos[contacto_encontrado_key]
                                print(f"Nombre de contacto actualizado de '{contacto_encontrado_key}' a '{nuevo_nombre}'.\n")
                                contacto_encontrado_key = nuevo_nombre # Actualizamos la clave para futuras modificaciones
                                break # Salimos del bucle de modificación
                        elif opcion_modificar == "2":
                            nuevo_email = input("• Introduce el NUEVO EMAIL: ").upper()
                            contactos[contacto_encontrado_key][0] = nuevo_email
                            print("Email actualizado correctamente.\n")
                            break # Salimos del bucle de modificación
                        elif opcion_modificar == "3":
                            nuevo_telefono = input("• Introduce el NUEVO TELEFONO: ").upper()
                            contactos[contacto_encontrado_key][1] = nuevo_telefono
                            print("Teléfono actualizado correctamente.\n")
                            break # Salimos del bucle de modificación
                        elif opcion_modificar == "4":
                            print("Modificación cancelada.\n")
                            break # Salimos del bucle de modificación
                        else:
                            print("Opción no válida. Por favor, elige 1, 2, 3 o 4.\n")
                else:
                    print(f"\n¡No se encontró ningún contacto con '{termino_busqueda}' en la agenda para modificar!\n")
            input("Presiona Enter para continuar...")

    elif agenda == "5":
            # ELIMINAR
            if not contactos:
                print("¡La agenda está vacía! No hay contactos para eliminar.")
            else:
                termino_busqueda = input('• Introduce el NOMBRE, EMAIL o TELEFONO del contacto a ELIMINAR. . .').upper()
                encontrado = False
                contactos_a_eliminar = [] # Usamos una lista para almacenar las claves a eliminar

                for nombre_contacto, datos_contacto in contactos.items():
                    email = datos_contacto[0]
                    telefono = datos_contacto[1]

                    if (termino_busqueda == nombre_contacto or
                        termino_busqueda == email or
                        termino_busqueda == telefono):
                        contactos_a_eliminar.append(nombre_contacto) # Agregamos el nombre del contacto a la lista
                        encontrado = True

                if encontrado:
                    for nombre_contacto_eliminar in contactos_a_eliminar:
                        del contactos[nombre_contacto_eliminar] # Eliminamos el contacto del diccionario
                        print(f"El contacto '{nombre_contacto_eliminar}' ha sido ELIMINADO de la agenda.\n")
                else:
                    print(f"\n¡No se encontró ningún contacto con '{termino_busqueda}' en la agenda para eliminar!\n")
            input("Presiona Enter para continuar...")

    elif agenda == "6":
        # SALIR
        salir += True

```
